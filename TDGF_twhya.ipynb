## Continuation Feature Demonstration

The `fit_disk_pipeline` function now supports **continuation** from previously saved MCMC samplers. This is useful when you want to:

1. **Extend a previous run** with additional steps
2. **Resume interrupted computations**
3. **Refine results** by running more MCMC steps

### Key Features:

- **`continue_from_sampler`**: Path to an existing `.pkl` sampler file
- **`nsteps`**: Number of **ADDITIONAL** steps to run (not total)
- **Validation**: Automatically checks compatibility (walker count, dimensions)
- **Clear logging**: Shows whether running in "CONTINUATION MODE" or "FRESH RUN MODE"

### Usage:

To continue from an existing sampler, add this to your TOML configuration:

```toml
[mcmc_settings]
continue_from_sampler = "path/to/existing/sampler.pkl"
nsteps = 1000  # Additional steps to run
# ... other settings ...
```
